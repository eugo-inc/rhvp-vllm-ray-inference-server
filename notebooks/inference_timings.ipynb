# vLLM + Ray Serve inference server: timings

This notebook times the inference server this pattern installs: stock vLLM served by Ray Serve LLM, built only from stock packages. It talks to the server inside the cluster through its OpenAI-compatible API. Each section measures the server live.

Run the cells from the top. The first cell waits until the server answers a one-token completion. To check the layout without a GPU, start JupyterLab with RHVP_REPLAY set to a recording from the recordings folder.

```python
from rhvp_bench import run

run.ready()
```

## 1 · Server start to ready

How long the server pod takes from being placed on a node to Ready, read from the pod's own Kubernetes conditions. The pod turns Ready once the server has loaded its model and is serving. One start of the pod is one measurement.

```python
run.startup()
```

## 2 · Time to first token

The time from just before a request is sent to the first generated text coming back, for requests sent one at a time. Every request starts with a fresh random tag, and vLLM's prefix cache (on by default) has nothing to reuse.

```python
run.first_token()
```

## 3 · Decode speed (one stream)

How fast one stream generates text once it has started: the tokens after the first group of streamed tokens, divided by the time from the first generated text to the last. Requests are sent one at a time.

```python
run.decode()
```

## 4 · Throughput at 1, 8 and 32 concurrent streams

All generated tokens per second, with 1, 8 and 32 streams running at the same time, every stream generating the same number of tokens.

```python
run.throughput()
```
